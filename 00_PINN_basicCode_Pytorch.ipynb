```python
import torch
import torch.nn as nn
```

#Create Neural Network

```python
model = nn.Sequential(
    nn.Linear(1, 1),
    nn.Tanh(),
    nn.Linear(1, 1)
)
```

```python
model
```

# Set Weights

```python
with torch.no_grad():

  model[0].weight.fill_(2.0)
  model[0].bias.fill_(1.0)

  model[2].weight.fill_(3.0)
  model[2].bias.fill_(5.0)
```

Create Input t=0.5

```python
t = torch.tensor([[0.5]], requires_grad=True)
```

```python
t
```

# Forward Pass

```python
T=model(t)
```

```python
print(T)
```

# derivative dT/dt

```python
dT_dt=torch.autograd.grad(
    outputs=T,
    inputs=t,
    grad_outputs=torch.ones_like(T),
    create_graph=True
)[0]
```

```python
print("dT_dt",dT_dt)
```

#Residual

```python
residual=dT_dt+0.1*(T-25)
print("REsidual",residual)
```

#Physics Loss

```python
physics_loss=torch.mean(residual**2)
physics_loss
```

# Initial Condition

```python
t0=torch.tensor([[0.0]],requires_grad=True)
T0=model(t0)
print(T0)
```

#Initial Condition LOss

```python
ic_loss=torch.mean((T0-100)**2)
ic_loss
```

#Total_loss

```python
total_loss=physics_loss+ic_loss
print(total_loss)
```

#Backpropagation

```python
total_loss.backward()
```

```python
print(model[0].weight.grad)
print(model[0].bias.grad)

print(model[2].weight.grad)
print(model[2].bias.grad)
```
