```python
import torch
```

```python
x = torch.tensor(
    [[0.5]],
    dtype=torch.float32,
    requires_grad=True
)
x.shape
```

```python
t = torch.tensor(
    [[0.2]],
    dtype=torch.float32,
    requires_grad=True
)
t
```

# Simple Neural Network

```python
z=2*x+3*t+1
h=torch.tanh(z)
T=4*h+5

print("Z: ",z)
print("h: ",h)
print("T: ",T)
```

# dT/dt

```python
T_t=torch.autograd.grad(
    outputs=T,
    inputs=t,
    grad_outputs=torch.ones_like(T),
    create_graph=True)[0]
T_t
```

# dT/dx

```python
T_x=torch.autograd.grad(
    outputs=T,
    inputs=x,
    grad_outputs=torch.ones_like(T),
    create_graph=True
)[0]
T_x
```

#d^2T/dx^2

```python
T_xx=torch.autograd.grad(
  outputs=T_x,
  inputs=x,
  grad_outputs=torch.ones_like(T_x),
  create_graph=True
)[0]
T_xx
```

#Physics Residual

```python
alpha=0.01

residual=T_t-alpha*T_xx
print("residual : ",residual)
```

```python
physics_loss=torch.mean(residual**2)
physics_loss
```
